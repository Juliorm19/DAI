### Clases Celdas y Tablero

```python
class Celdas:
    RESET = "\033[0m"
    NEGRITA = "\033[1m"
    TEXTO_NEGRO = "\033[30m"

    # Colores de fondo ANSI para consola
    FONDO_GRIS = "\033[48;5;240m"   # Fuera del puzzle
    FONDO_NEGRO = "\033[40m"        # Usada / Bloqueada / Lokmon
    FONDO_BLANCO = "\033[107m"      # Disponible

    terrenos = {
        "vacio": f"{FONDO_GRIS}   {RESET}",
        "bloqueada": f"{FONDO_NEGRO}   {RESET}",
        "libre": f"{FONDO_BLANCO}   {RESET}",
    }

    simbolos = {
        "L": f"{FONDO_BLANCO}{TEXTO_NEGRO}{NEGRITA} L {RESET}",
        "T": f"{FONDO_BLANCO}{TEXTO_NEGRO}{NEGRITA} T {RESET}",
        "A": f"{FONDO_BLANCO}{TEXTO_NEGRO}{NEGRITA} A {RESET}",
        "K": f"{FONDO_BLANCO}{TEXTO_NEGRO}{NEGRITA} K {RESET}",
        "X": f"{FONDO_BLANCO}{TEXTO_NEGRO}{NEGRITA} X {RESET}",
        "O": f"{FONDO_BLANCO}{TEXTO_NEGRO}{NEGRITA} O {RESET}",
        "R": f"{FONDO_BLANCO}{TEXTO_NEGRO}{NEGRITA} R {RESET}",
        "E": f"{FONDO_BLANCO}{TEXTO_NEGRO}{NEGRITA} E {RESET}",
        "Lokmon": f"{FONDO_NEGRO}   {RESET}"
    }

    def __init__(self, terreno="vacio", contenido=None):
        self.terreno = terreno
        self.contenido = contenido

    def __str__(self) -> str:
        if self.contenido in self.simbolos:
            return self.simbolos[self.contenido]
        return self.terrenos.get(self.terreno, "❓")
```

```python
class Tablero:

    def __init__(self, filas=10, columnas=10):
        self.filas = filas
        self.columnas = columnas
        self.grid = np.array(
            [
                [Celdas(terreno="vacio") for _ in range(columnas)]
                for _ in range(filas)
            ],
            dtype=object,
        )
        # Variables de estado del motor de juego
        self.fase = "NORMAL"  # "NORMAL", "COLOCAR_Lokmon", "SELECCIONAR_OBJETIVO_TA"
        self.Lokmones_pendientes = 0
        self.primer_Lokmon_pos = None
        self.requiere_adyacencia = False

    def cargar_nivel_desde_txt(
        self, ruta_archivo: str = None, nivel_id: int = 1
    ) -> bool:
        if ruta_archivo is None:
            ruta_archivo = f"nivel_{nivel_id}.txt"

        try:
            with open(ruta_archivo, mode="r", encoding="utf-8") as archivo:
                lineas = [linea.strip() for linea in archivo if linea.strip()]

            filas = 0
            columnas = 0
            lineas_mapa = []
            leyendo_mapa = False

            for linea in lineas:
                if linea.startswith("filas:"):
                    filas = int(linea.split(":")[1].strip())
                elif linea.startswith("columnas:"):
                    columnas = int(linea.split(":")[1].strip())
                elif linea.startswith("mapa:"):
                    leyendo_mapa = True
                elif leyendo_mapa:
                    lineas_mapa.append(linea)

            self.filas = filas
            self.columnas = columnas
            self.grid = np.array(
                [
                    [Celdas(terreno="vacio") for _ in range(self.columnas)]
                    for _ in range(self.filas)
                ],
                dtype=object,
            )

            for f, f_str in enumerate(lineas_mapa):
                if f >= self.filas:
                    break
                f_str_limpia = (
                    f_str.replace("–", "-").replace("—", "-").rstrip(";")
                )
                celdas_str = f_str_limpia.split()

                for c, char in enumerate(celdas_str):
                    if c >= self.columnas or char == "-":
                        continue

                    self.grid[f, c].terreno = "libre"
                    if char == "_":
                        self.grid[f, c].contenido = None
                    else:
                        self.grid[f, c].contenido = char

            return True

        except Exception as e:
            print(f"Error al procesar el archivo '{ruta_archivo}': {e}")
            return False

    def copiar(self):
        """Genera una copia profunda del estado del tablero para los nodos del grafo."""
        nuevo_tablero = copy.deepcopy(self)
        return nuevo_tablero

    def obtener_firma(self) -> tuple:
        """Crea una tupla inmutable con el estado completo del tablero."""
        matriz_estado = tuple(
            tuple((c.terreno, c.contenido) for c in fila) for fila in self.grid
        )

        return (
            matriz_estado,
            self.fase,
            self.Lokmones_pendientes,
            self.primer_Lokmon_pos,
            self.requiere_adyacencia,
        )

    def verificar_victoria(self) -> bool:
        """Comprueba si el tablero está libre de letras pendientes y no quedan Lokmones por colocar."""
        if self.Lokmones_pendientes > 0:
            return False
        for fil in range(self.filas):
            for col in range(self.columnas):
                celda = self.grid[fil, col]
                if celda.terreno == "libre" and celda.contenido != "Lokmon":
                    return False
        return True

    def consumir_casilla(self, f: int, c: int):
        celda = self.grid[f, c]
        if celda.contenido != "X":
            celda.contenido = None
            celda.terreno = "bloqueada"

    def es_linea_conectada(self, pos1: tuple, pos2: tuple) -> bool:
        f1, c1 = pos1
        f2, c2 = pos2

        if f1 != f2 and c1 != c2:
            return False

        paso_c = 1 if c2 > c1 else (-1 if c2 < c1 else 0)
        paso_f = 1 if f2 > f1 else (-1 if f2 < f1 else 0)

        f_actual, c_actual = f1 + paso_f, c1 + paso_c
        while (f_actual, c_actual) != (f2, c2):
            celda = self.grid[f_actual, c_actual]
            es_pasable = (celda.terreno == "bloqueada") or (
                celda.contenido in ["Lokmon", "X", "?"]
            )
            if not es_pasable:
                return False
            f_actual += paso_f
            c_actual += paso_c

        return True

    def validar_secuencia_adyacente(self, coordenadas: list) -> bool:
        if len(coordenadas) <= 1:
            return True
        for i in range(len(coordenadas) - 1):
            if not self.es_linea_conectada(coordenadas[i], coordenadas[i + 1]):
                return False
        return True

    def aplicar_jugada(
        self, coords: list, palabra: str = None, pos_objetivo: tuple = None
    ) -> bool:
        """Aplica una accion sobre la copia actual del tablero segun la fase de juego."""
        if self.fase == "NORMAL":
            casillas_a_consumir = []
            for f, c in coords:
                celda = self.grid[f, c]
                if celda.contenido and celda.contenido not in ["X", "Lokmon"]:
                    casillas_a_consumir.append((f, c))

            for f, c in casillas_a_consumir:
                self.consumir_casilla(f, c)

            if palabra == "LOK":
                self.Lokmones_pendientes = 1
                self.requiere_adyacencia = False
                self.fase = "COLOCAR_Lokmon"
            elif palabra == "TLAK":
                self.Lokmones_pendientes = 2
                self.primer_Lokmon_pos = None
                self.requiere_adyacencia = True
                self.fase = "COLOCAR_Lokmon"
            elif palabra == "TA":
                self.fase = "SELECCIONAR_OBJETIVO_TA"
            return True

        elif self.fase == "SELECCIONAR_OBJETIVO_TA":
            f, c = pos_objetivo
            celda_objetivo = self.grid[f, c]
            tipo_objetivo = celda_objetivo.contenido

            for fil in range(self.filas):
                for col in range(self.columnas):
                    celda = self.grid[fil, col]
                    if (
                        celda.terreno == "libre"
                        and celda.contenido == tipo_objetivo
                    ):
                        celda.contenido = "Lokmon"

            self.fase = "NORMAL"
            return True

        elif self.fase == "COLOCAR_Lokmon":
            f, c = pos_objetivo
            celda = self.grid[f, c]
            celda.contenido = "Lokmon"
            self.Lokmones_pendientes -= 1

            if self.requiere_adyacencia and self.primer_Lokmon_pos is None:
                self.primer_Lokmon_pos = (f, c)

            if self.Lokmones_pendientes == 0:
                self.fase = "NORMAL"
                self.requiere_adyacencia = False
                self.primer_Lokmon_pos = None
            return True

        return False

    def __str__(self) -> str:
        grid_str = (
            "    "
            + "  ".join([str(c) for c in range(self.columnas)])
            + "\n"
        )
        for fila in range(self.filas):
            grid_str += f"{fila}  "
            for columna in range(self.columnas):
                grid_str += str(self.grid[fila, columna])
            grid_str += "\n"
        return grid_str
```

### Clase de los Nodos, incluye función de backtraking (obtener_caminos) y función para la generación de los sucesores

```python
class NodoBusqueda:
    def __init__(self, tablero, padre=None, accion=None, costo=0, h=0):
        self.tablero = (
            tablero
        )
        self.padre = padre
        self.accion = accion  # Cadena de coordenadas ejecutadas (ej: "3,3 3,4 3,5" o "4,3")
        self.g = costo  # Costo real acumulado g(n) = profundidad
        self.h = h
        self.f = self.g + self.h  # f(n) = g(n) + h(n) para A*

    def __lt__(self, otro):
        #Permite ordenar los nodos en la Cola de Prioridad (heapq) según f(n).
        return self.f < otro.f

    def obtener_camino(self):
        #Reconstruye la lista de acciones desde la raíz hasta este nodo.
        camino = []
        nodo_actual = self
        while nodo_actual.padre is not None:
            camino.append(nodo_actual.accion)
            nodo_actual = nodo_actual.padre
        camino.reverse()
        return camino


def resolver_comodin_texto(palabra_bruta: str) -> str:
    if len(palabra_bruta) == 3:
        if all(p == c or p == "?" for p, c in zip(palabra_bruta, "LOK")):
            return "LOK"
    elif len(palabra_bruta) == 4:
        if all(p == c or p == "?" for p, c in zip(palabra_bruta, "TLAK")):
            return "TLAK"
    elif len(palabra_bruta) == 2:
        if all(p == c or p == "?" for p, c in zip(palabra_bruta, "TA")):
            return "TA"
    return palabra_bruta

def calcular_heuristica(tablero):
    h = 0

    # 1. Costos exactos y obligatorios
    h += tablero.Lokmones_pendientes

    if tablero.fase == "SELECCIONAR_OBJETIVO_TA":
        h += 1

    # 2. Estimación del costo por letras restantes
    letras_restantes = 0
    for f in range(tablero.filas):
        for c in range(tablero.columnas):
            celda = tablero.grid[f, c]
            # Contamos casillas libres que contengan una letra válida o comodín
            # (Ignoramos las 'X' porque son puentes infinitos y los 'Lokmon' porque ya están resueltos)
            if celda.terreno == "libre" and celda.contenido not in [None, "X", "Lokmon"]:
                letras_restantes += 1

    # En LOK, la palabra más larga que consume letras de golpe es "TLAK" (4 letras).
    # En el mejor escenario absoluto (pura eficiencia), borraríamos 4 letras por turno.
    h += math.ceil(letras_restantes / 4)

    return h


def generar_sucesores(nodo_padre: NodoBusqueda, heuristica=None) -> list:
    if heuristica is None:
        heuristica = calcular_heuristica

    sucesores = []
    tablero = nodo_padre.tablero

    # FASE 1: BUSCAR PALABRAs
    if tablero.fase == "NORMAL":
        direcciones = [(0, 1), (1, 0), (0, -1), (-1, 0)]

        for f in range(tablero.filas):
            for c in range(tablero.columnas):
                celda_origen = tablero.grid[f, c]

                # Debe empezar obligatoriamente en una letra o comodin activo
                if celda_origen.terreno != "libre" or not celda_origen.contenido or celda_origen.contenido in ["X", "Lokmon"]:
                    continue

                for df, dc in direcciones:
                    for longitud in range(2, max(tablero.filas, tablero.columnas) + 1):
                        coords = []
                        letras = []
                        valido = True

                        for i in range(longitud):
                            nf, nc = f + i * df, c + i * dc
                            if not (0 <= nf < tablero.filas and 0 <= nc < tablero.columnas):
                                valido = False
                                break

                            celda = tablero.grid[nf, nc]

                            # 1. ¿Es un puente permitido? (X, Lokmon o casilla ya consumida)
                            es_puente = (celda.terreno == "bloqueada") or (celda.contenido in ["X", "Lokmon"])

                            # 2. ¿Es una letra activa? (libre y con contenido)
                            es_letra = (celda.terreno == "libre") and (celda.contenido is not None) and (celda.contenido not in ["X", "Lokmon"])

                            # SI NO ES PUENTE NI ES LETRA ACTIVA (ej: es un espacio libre vacio '_'), ES ILEGAL
                            if not es_puente and not es_letra:
                                valido = False
                                break

                            coords.append((nf, nc))

                            if es_letra:
                                letras.append(str(celda.contenido))

                        if not valido or not letras:
                            continue

                        palabra_bruta = "".join(letras)
                        palabra_evaluada = resolver_comodin_texto(palabra_bruta)

                        if palabra_evaluada in ["LOK", "TLAK", "TA"]:
                            # Formatear la salida asignando únicamente el contenido real
                            accion_str = " ".join(f"{rf},{rc}({tablero.grid[rf, rc].contenido})" for rf, rc in coords)

                            nuevo_tablero = tablero.copiar()
                            nuevo_tablero.aplicar_jugada(coords, palabra=palabra_evaluada)
                            h_calculada = heuristica(nuevo_tablero)

                            hijo = NodoBusqueda(
                                tablero=nuevo_tablero,
                                padre=nodo_padre,
                                accion=accion_str,
                                costo=nodo_padre.g + 1,
                                h=h_calculada
                            )
                            sucesores.append(hijo)

    # FASE 2: COLOCAR Lokmon
    elif tablero.fase == "COLOCAR_Lokmon":
        for f in range(tablero.filas):
            for c in range(tablero.columnas):
                celda = tablero.grid[f, c]
                # Los Lokmones se pueden colocar en casillas libres o sobre casillas con 'X' / bloqueadas de puente
                if (celda.terreno == "libre" and celda.contenido != "Lokmon") or celda.contenido == "X":
                    if tablero.requiere_adyacencia and tablero.primer_Lokmon_pos is not None:
                        f1, c1 = tablero.primer_Lokmon_pos
                        es_conectado = (abs(f1 - f) + abs(c1 - c) == 1) or tablero.es_linea_conectada((f1, c1), (f, c))
                        if not es_conectado:
                            continue

                    nuevo_tablero = tablero.copiar()
                    nuevo_tablero.aplicar_jugada([], pos_objetivo=(f, c))
                    h_calculada = heuristica(nuevo_tablero)

                    hijo = NodoBusqueda(
                        tablero=nuevo_tablero,
                        padre=nodo_padre,
                        accion=f"{f},{c}(Lokmon)",
                        costo=nodo_padre.g + 1,
                        h=h_calculada
                    )
                    sucesores.append(hijo)

    # FASE 3: APLICAR TA
    elif tablero.fase == "SELECCIONAR_OBJETIVO_TA":
        tipos_evaluados = set()
        for f in range(tablero.filas):
            for c in range(tablero.columnas):
                celda = tablero.grid[f, c]
                if celda.terreno == "libre" and celda.contenido != "Lokmon":
                    if celda.contenido not in tipos_evaluados:
                        tipos_evaluados.add(celda.contenido)
                        nuevo_tablero = tablero.copiar()
                        nuevo_tablero.aplicar_jugada([], pos_objetivo=(f, c))
                        h_calculada = heuristica(nuevo_tablero)

                        hijo = NodoBusqueda(
                            tablero=nuevo_tablero,
                            padre=nodo_padre,
                            accion=f"{f},{c}(Uno o varios Lokmons)",
                            costo=nodo_padre.g + 1,
                            h=h_calculada
                        )
                        sucesores.append(hijo)

    return sucesores
```

### Clase de la frontera para los diferentes algoritmos de busqueda

```python
class StackFrontier:
    #Frontera LIFO para Búsqueda en Profundidad (DFS)
    def __init__(self):
        self.frontera = []

    def add(self, nodo):
        self.frontera.append(nodo)

    def empty(self) -> bool:
        return len(self.frontera) == 0

    def remove(self):
        if self.empty():
            raise Exception("Frontera vacía")
        return self.frontera.pop()


class QueueFrontier(StackFrontier):
    #Frontera FIFO para Búsqueda en Anchura (BFS)
    def remove(self):
        if self.empty():
            raise Exception("Frontera vacía")
        return self.frontera.pop(0)


class PriorityQueueFrontier:
    #Frontera ordenada por f(n) para Búsqueda Informada (A*)
    def __init__(self):
        self.frontera = []

    def add(self, nodo):
        heapq.heappush(self.frontera, nodo)

    def empty(self) -> bool:
        return len(self.frontera) == 0

    def remove(self):
        if self.empty():
            raise Exception("Frontera vacía")
        return heapq.heappop(self.frontera)
```

### Algoritmo BFS de prueba

```python
def busqueda_bfs(tablero_inicial):
    tracemalloc.start()
    t_inicio = time.perf_counter()

    nodo_raiz = NodoBusqueda(tablero=tablero_inicial, costo=0)

    if tablero_inicial.verificar_victoria():
        t_fin = time.perf_counter()
        _, ram_peak = tracemalloc.get_traced_memory()
        tracemalloc.stop()
        return nodo_raiz, 0, 0, 0, t_fin - t_inicio, ram_peak / (1024 * 1024)

    # Uso explicito de QueueFrontier
    frontera = QueueFrontier()
    frontera.add(nodo_raiz)

    explorados = set()
    explorados.add(tablero_inicial.obtener_firma())

    nodos_expandidos = 0
    max_profundidad = 0

    while not frontera.empty():
        nodo_actual = frontera.remove()

        if nodo_actual.tablero.verificar_victoria():
            t_fin = time.perf_counter()
            _, ram_peak = tracemalloc.get_traced_memory()
            tracemalloc.stop()
            return (
                nodo_actual,
                nodos_expandidos,
                nodo_actual.g,
                max_profundidad,
                t_fin - t_inicio,
                ram_peak / (1024 * 1024),
            )

        sucesores = generar_sucesores(nodo_actual)
        nodos_expandidos += 1

        for nodo_hijo in sucesores:
            firma = nodo_hijo.tablero.obtener_firma()

            if firma not in explorados:
                explorados.add(firma)
                frontera.add(nodo_hijo)

                max_profundidad = max(
                    max_profundidad,
                    nodo_hijo.g
                )

    t_fin = time.perf_counter()
    _, ram_peak = tracemalloc.get_traced_memory()
    tracemalloc.stop()
    return None, nodos_expandidos, 0, max_profundidad, t_fin - t_inicio, ram_peak / (1024 * 1024)
```

#Algoritmo DFS

```python
def busqueda_dfs(tablero_inicial):
    # Iniciar medición de memoria y tiempo
    tracemalloc.start()
    t_inicio = time.perf_counter()

    # Nodo raíz
    nodo_raiz = NodoBusqueda(
        tablero=tablero_inicial,
        costo=0
    )

    # Si el nivel ya está resuelto desde el inicio
    if tablero_inicial.verificar_victoria():
        t_fin = time.perf_counter()
        _, ram_peak = tracemalloc.get_traced_memory()
        tracemalloc.stop()

        return (
            nodo_raiz,
            0,
            0,
            0,
            t_fin - t_inicio,
            ram_peak / (1024 * 1024)
        )

    # DFS utiliza una pila (LIFO)
    frontera = StackFrontier()
    frontera.add(nodo_raiz)

    # Estados ya explorados
    explorados = set()

    # Firmas de estados que actualmente están dentro de la frontera
    firmas_en_frontera = {
        tablero_inicial.obtener_firma()
    }

    # Métricas
    nodos_expandidos = 0
    max_profundidad = 0

    while not frontera.empty():

        # Sacar el último nodo agregado a la pila
        nodo_actual = frontera.remove()

        firma_actual = nodo_actual.tablero.obtener_firma()

        # Ya no está pendiente en la frontera
        firmas_en_frontera.discard(firma_actual)

        # Si ya había sido explorado, no volver a procesarlo
        if firma_actual in explorados:
            continue

        # Marcar estado como explorado
        explorados.add(firma_actual)

        # Comprobar si es estado objetivo
        if nodo_actual.tablero.verificar_victoria():
            t_fin = time.perf_counter()
            _, ram_peak = tracemalloc.get_traced_memory()
            tracemalloc.stop()

            return (
                nodo_actual,
                nodos_expandidos,
                nodo_actual.g,
                max_profundidad,
                t_fin - t_inicio,
                ram_peak / (1024 * 1024)
            )

        # Generar sucesores del estado actual
        sucesores = generar_sucesores(nodo_actual)

        # El nodo se cuenta como expandido porque generamos sus sucesores
        nodos_expandidos += 1

        # DFS usa pila.
        # Se recorren los sucesores al revés para conservar
        # el orden natural generado por generar_sucesores().
        for nodo_hijo in reversed(sucesores):

            firma_hijo = nodo_hijo.tablero.obtener_firma()

            # Equivalente a:
            # neighbor not in frontier U explored
            if (
                firma_hijo not in explorados
                and firma_hijo not in firmas_en_frontera
            ):
                frontera.add(nodo_hijo)
                firmas_en_frontera.add(firma_hijo)

                # Actualizar profundidad máxima alcanzada
                max_profundidad = max(
                    max_profundidad,
                    nodo_hijo.g
                )

    # Si se vacía la frontera sin encontrar solución
    t_fin = time.perf_counter()
    _, ram_peak = tracemalloc.get_traced_memory()
    tracemalloc.stop()

    return (
        None,
        nodos_expandidos,
        0,
        max_profundidad,
        t_fin - t_inicio,
        ram_peak / (1024 * 1024)
    )
```

#Algoritmo A*

```python

# ============================================================
# TRES HEURÍSTICAS
# ============================================================

def heuristica_h1_acciones_minimas(tablero):
    """
    H1: heurística aplicada por Claiby
    """
    return calcular_heuristica(tablero)


def _letras_restantes_lok(tablero):
    total = 0

    for f in range(tablero.filas):
        for c in range(tablero.columnas):
            celda = tablero.grid[f, c]

            if (
                celda.terreno == "libre"
                and celda.contenido not in [None, "X", "Lokmon"]
            ):
                total += 1

    return total


def _max_palabra_disponible_lok(tablero):
    """
    Busca de forma segura el mayor número de letras que puede consumir
    una palabra válida actualmente disponible.
    """
    if tablero.fase != "NORMAL":
        return 4

    mejor = 0
    direcciones = [(0, 1), (1, 0), (0, -1), (-1, 0)]
    max_longitud = max(tablero.filas, tablero.columnas)

    for f in range(tablero.filas):
        for c in range(tablero.columnas):

            origen = tablero.grid[f, c]

            if (
                origen.terreno != "libre"
                or not origen.contenido
                or origen.contenido in ["X", "Lokmon"]
            ):
                continue

            for df, dc in direcciones:

                for longitud in range(2, max_longitud + 1):

                    letras = []
                    valido = True

                    for i in range(longitud):

                        nf = f + i * df
                        nc = c + i * dc

                        if not (
                            0 <= nf < tablero.filas
                            and 0 <= nc < tablero.columnas
                        ):
                            valido = False
                            break

                        celda = tablero.grid[nf, nc]

                        es_puente = (
                            celda.terreno == "bloqueada"
                            or celda.contenido in ["X", "Lokmon"]
                        )

                        es_letra = (
                            celda.terreno == "libre"
                            and celda.contenido is not None
                            and celda.contenido not in ["X", "Lokmon"]
                        )

                        if not es_puente and not es_letra:
                            valido = False
                            break

                        if es_letra:
                            letras.append(str(celda.contenido))

                    if not valido or not letras:
                        continue

                    palabra = resolver_comodin_texto("".join(letras))

                    if palabra in ["TA", "LOK", "TLAK"]:
                        mejor = max(mejor, len(letras))

    return mejor


def heuristica_h2_eficiencia(tablero):
    """
    H2: considera la capacidad de consumo de una palabra disponible.

    Si actualmente existe TLAK, el estado tiene capacidad inmediata
    de consumir 4 letras; si no, se adapta al mejor movimiento disponible.
    """
    h = tablero.Lokmones_pendientes

    if tablero.fase == "SELECCIONAR_OBJETIVO_TA":
        h += 1

    letras = _letras_restantes_lok(tablero)

    if letras == 0:
        return h

    mejor_consumo = _max_palabra_disponible_lok(tablero)

    if mejor_consumo <= 0:
        mejor_consumo = 4

    h += math.ceil(letras / min(4, mejor_consumo))

    return h


def heuristica_h3_potencial(tablero):
    """
    H3: evalúa el potencial actual del tablero.

    Da prioridad a estados que poseen oportunidades de mayor consumo.
    """
    h = tablero.Lokmones_pendientes

    if tablero.fase == "SELECCIONAR_OBJETIVO_TA":
        h += 1

    letras = _letras_restantes_lok(tablero)

    if letras == 0:
        return h

    mejor_consumo = _max_palabra_disponible_lok(tablero)

    # Potencial inmediato:
    # TLAK = 4, LOK = 3, TA = 2.
    if mejor_consumo >= 4:
        capacidad = 4
    elif mejor_consumo == 3:
        capacidad = 3
    elif mejor_consumo == 2:
        capacidad = 2
    else:
        capacidad = 4

    h += math.ceil(letras / capacidad)

    return h


# Mantiene disponible el nombre original.
calcular_heuristica_original = calcular_heuristica
```

```python
def busqueda_a_star(tablero_inicial, heuristica):
    """
    A* con heurística seleccionable.
    """
    tracemalloc.start()
    t_inicio = time.perf_counter()

    h_inicial = heuristica(tablero_inicial)

    nodo_raiz = NodoBusqueda(
        tablero=tablero_inicial,
        costo=0,
        h=h_inicial
    )

    if tablero_inicial.verificar_victoria():
        t_fin = time.perf_counter()
        _, ram_peak = tracemalloc.get_traced_memory()
        tracemalloc.stop()

        return (
            nodo_raiz,
            0,
            0,
            0,
            t_fin - t_inicio,
            ram_peak / (1024 * 1024)
        )

    frontera = PriorityQueueFrontier()
    frontera.add(nodo_raiz)

    costos_g = {
        tablero_inicial.obtener_firma(): 0
    }

    nodos_expandidos = 0
    max_profundidad = 0

    while not frontera.empty():

        nodo_actual = frontera.remove()

        if nodo_actual.g > max_profundidad:
            max_profundidad = nodo_actual.g

        if nodo_actual.tablero.verificar_victoria():

            t_fin = time.perf_counter()
            _, ram_peak = tracemalloc.get_traced_memory()
            tracemalloc.stop()

            return (
                nodo_actual,
                nodos_expandidos,
                nodo_actual.g,
                max_profundidad,
                t_fin - t_inicio,
                ram_peak / (1024 * 1024)
            )

        sucesores = generar_sucesores(
            nodo_actual,
            heuristica=heuristica
        )

        nodos_expandidos += 1

        for nodo_hijo in sucesores:

            firma = nodo_hijo.tablero.obtener_firma()

            if (
                firma not in costos_g
                or nodo_hijo.g < costos_g[firma]
            ):
                costos_g[firma] = nodo_hijo.g
                frontera.add(nodo_hijo)

    t_fin = time.perf_counter()
    _, ram_peak = tracemalloc.get_traced_memory()
    tracemalloc.stop()

    return (
        None,
        nodos_expandidos,
        0,
        max_profundidad,
        t_fin - t_inicio,
        ram_peak / (1024 * 1024)
    )
```

### Exportador de Métricas

```python
def guardar_output_algoritmo(
    nombre_archivo,
    nodo_meta,
    nodos_expandidos,
    profundidad_solucion,
    max_profundidad,
    tiempo_ejecucion,
    ram_consumida
):
    ruta_completa = os.path.join(os.getcwd(), nombre_archivo)

    if nodo_meta is not None:
        path_to_goal = nodo_meta.obtener_camino()
        cost_of_path = len(path_to_goal)
        search_depth = profundidad_solucion
    else:
        path_to_goal = []
        cost_of_path = 0
        search_depth = 0

    contenido = (
        f"path_to_goal: {path_to_goal}\n"
        f"cost_of_path: {cost_of_path}\n"
        f"nodes_expanded: {nodos_expandidos}\n"
        f"search_depth: {search_depth}\n"
        f"max_search_depth: {max_profundidad}\n"
        f"running_time: {tiempo_ejecucion}\n"
        f"max_ram_usage: {ram_consumida:.6f}"
    )
    print(contenido)

    try:
        with open(ruta_completa, "w", encoding="utf-8") as f:
            f.write(contenido)
        print(f"Documento '{nombre_archivo}'\n")
    except Exception as e:
        print(f"No se pudo guardar '{nombre_archivo}': {e}\n")
```

```python
class JuegoLOKConsole:

    def __init__(self, nivel=1):
        self.nivel_actual = nivel
        self.tablero = Tablero()
        self.tablero.cargar_nivel_desde_txt(nivel_id=nivel)
        self.estado_juego = "NORMAL"
        self.Lokmones_pendientes = 0
        self.primer_Lokmon_pos = None
        self.requiere_adyacencia = False
        self.mensaje_estado = f"¡Nivel {nivel} cargado! Ingresa coordenadas para formar palabras."

    def __mostrar_tablero(self):
        """Limpia la pantalla e imprime el estado actual del tablero."""
        output.clear()
        print(f"      LOK GAME - NIVEL {self.nivel_actual}")
        print(self.tablero)
        print(f"ESTADO: {self.mensaje_estado}\n")

    def procesar_normal(self, entrada: str):
        try:
            coords = [tuple(map(int, par.split(','))) for par in entrada.strip().split()]

            # 1. Validar conexión tramo a tramo (soporta giros de 90° pasando por 'X')
            if not self.tablero.validar_secuencia_adyacente(coords):
                self.mensaje_estado = "Las celdas no están conectadas en línea recta o vía 'X'/'?'."
                return

            letras = []
            casillas_a_consumir = []

            # 2. Interpretar la secuencia
            for f, c in coords:
                celda = self.tablero.grid[f, c]
                contenido = celda.contenido

                if celda.terreno == "bloqueada":
                    self.mensaje_estado = f"La casilla ({f},{c}) está desactivada."
                    return

                # La 'X' actúa como conductor dentro del camino ingresado (no suma letra ni se consume)
                if contenido == "X":
                    continue

                # Comodín '?'
                elif contenido == "?":
                    letras.append("?")
                    casillas_a_consumir.append((f, c))

                elif contenido and contenido != "Lokmon":
                    letras.append(contenido)
                    casillas_a_consumir.append((f, c))
                else:
                    self.mensaje_estado = f"Casilla ({f},{c}) no contiene una letra o conductor válido."
                    return

            # 3. Mapeo de palabra
            palabra_bruta = "".join(letras)
            palabra = self._resolver_comodin(palabra_bruta)

            if palabra in ["LOK", "TLAK", "TA"]:
                # Consumir solo las letras normales/comodines (las 'X' se mantienen intactas)
                for f, c in casillas_a_consumir:
                    self.tablero.consumir_casilla(f, c)

                if palabra == "LOK":
                    self.Lokmones_pendientes = 1
                    self.requiere_adyacencia = False
                    self.estado_juego = "COLOCAR_Lokmon"
                    self.mensaje_estado = "¡LOK! Coloca 1 Lokmon (puedes usar casillas libres o casillas con 'X')."

                elif palabra == "TLAK":
                    self.Lokmones_pendientes = 2
                    self.primer_Lokmon_pos = None
                    self.requiere_adyacencia = True
                    self.estado_juego = "COLOCAR_Lokmon"
                    self.mensaje_estado = "¡TLAK! Coloca el 1º Lokmon (el 2º deberá ser adyacente)."

                elif palabra == "TA":
                    self.estado_juego = "SELECCIONAR_OBJETIVO_TA"
                    self.mensaje_estado = "¡TA formado! Elige la casilla tipo para tus Lokmons."
            else:
                self.mensaje_estado = f"'{palabra_bruta}' no forma una palabra válida (LOK, TLAK, TA)."

        except Exception:
            self.mensaje_estado = "Formato inválido. Ejemplo: 3,3 3,4 3,5"

    def _resolver_comodin(self, palabra: str) -> str:
        """Resuelve el comodín '?' según encaje en LOK, TLAK o TA."""
        if len(palabra) == 3:
            # Evaluar si encaja con LOK (ej. "?OK", "L?K", "LO?")
            if all(p == c or p == "?" for p, c in zip(palabra, "LOK")):
                return "LOK"
        elif len(palabra) == 4:
            # Evaluar si encaja con TLAK (ej. "T?AK", "TLA?")
            if all(p == c or p == "?" for p, c in zip(palabra, "TLAK")):
                return "TLAK"
        elif len(palabra) == 2:
            # Evaluar si encaja con TA (ej. "?A", "T?")
            if all(p == c or p == "?" for p, c in zip(palabra, "TA")):
                return "TA"
        return palabra

    def procesar_ta_objetivo(self, entrada: str):
        try:
            f, c = map(int, entrada.strip().split(','))
            celda_objetivo = self.tablero.grid[f, c]

            if celda_objetivo.terreno != "libre" or celda_objetivo.contenido == "Lokmon":
                self.mensaje_estado = f"Casilla ({f},{c}) no es válida."
                return

            #if celda_objetivo.contenido == "X":
                #self.mensaje_estado = "La letra 'X' no se puede usar con TA."
                #return

            tipo_objetivo = celda_objetivo.contenido
            colocados = 0

            for fil in range(self.tablero.filas):
                for col in range(self.tablero.columnas):
                    celda = self.tablero.grid[fil, col]
                    if celda.terreno == "libre" and celda.contenido == tipo_objetivo:
                        celda.contenido = "Lokmon"
                        colocados += 1

            self.estado_juego = "NORMAL"
            tipo_txt = tipo_objetivo if tipo_objetivo else "espacios libres"

            if self.tablero.verificar_victoria():
                self.mensaje_estado = f"¡VICTORIA! Se colocaron {colocados} Lokmon en '{tipo_txt}'."
            else:
                self.mensaje_estado = f"¡TA aplicado! Se colocaron {colocados} Lokmon en '{tipo_txt}'."

        except Exception:
            self.mensaje_estado = "Formato inválido. Ejemplo: 5,5"

    def procesar_Lokmon(self, entrada: str):
        try:
            f, c = map(int, entrada.strip().split(','))
            celda = self.tablero.grid[f, c]

            # REGLA: Puedes colocar un Lokmon en cualquier casilla jugable activa ("libre")
            # que no tenga ya un Lokmon previamente.
            if celda.terreno == "libre" and celda.contenido != "Lokmon":

                # Validación de adyacencia para el 2º Lokmon de TLAK
                if self.requiere_adyacencia and self.primer_Lokmon_pos is not None:
                    if not self.tablero.es_linea_conectada(self.primer_Lokmon_pos, (f, c)):
                        self.mensaje_estado = f"El 2º Lokmon debe estar en línea conectada con el 1º ({self.primer_Lokmon_pos})."
                        return

                # Coloca el Lokmon (sobrescribe cualquier letra o contenido previo)
                celda.contenido = "Lokmon"
                self.Lokmones_pendientes -= 1

                if self.requiere_adyacencia and self.primer_Lokmon_pos is None:
                    self.primer_Lokmon_pos = (f, c)

                if self.Lokmones_pendientes > 0:
                    self.mensaje_estado = f"1º Lokmon colocado en ({f},{c}). Pon el 2º adyacente."
                else:
                    self.estado_juego = "NORMAL"
                    self.requiere_adyacencia = False
                    self.primer_Lokmon_pos = None

                    if self.tablero.verificar_victoria():
                        self.mensaje_estado = "¡VICTORIA! Tablero completado con éxito."
                    else:
                        self.mensaje_estado = f"Lokmon puesto en ({f},{c})."
            else:
                self.mensaje_estado = "Casilla no válida (está desactivada o ya tiene un Lokmon)."

        except Exception:
            self.mensaje_estado = "Formato inválido. Ejemplo: 4,4"

    def jugar(self) -> bool:
        """Bucle principal (main_loop). Retorna True si gana el nivel, False si sale."""
        while True:
            # 1. Mostrar tablero
            self.__mostrar_tablero()

            # 2. Condicional de salida (Victoria)
            if self.tablero.verificar_victoria():
                print(f"Has completado el Nivel {self.nivel_actual}.")
                input("\nPresiona ENTER para continuar al siguiente nivel...")
                return True

            # 3. Pedir entrada
            try:
                entrada = input("Ingresa tu jugada (o 's' / 'r'): ").strip()
            except (KeyboardInterrupt, EOFError):
                print("\nJuego cancelado.")
                return False

            # 4. Comandos del sistema
            if entrada.lower() == "s":
                print("Juego finalizado.")
                return False

            if entrada.lower() == "r":
                self.__init__(nivel=self.nivel_actual)
                continue

            if not entrada:
                continue

            # 5. Enrutamiento según el estado del juego
            if self.estado_juego == "NORMAL":
                self.procesar_normal(entrada)
            elif self.estado_juego == "SELECCIONAR_OBJETIVO_TA":
                self.procesar_ta_objetivo(entrada)
            elif self.estado_juego == "COLOCAR_Lokmon":
                self.procesar_Lokmon(entrada)
```
